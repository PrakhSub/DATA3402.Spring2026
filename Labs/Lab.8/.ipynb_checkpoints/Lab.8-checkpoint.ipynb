### Exercise 3: Training a Classifier

Lets' repeat what we did manually in the previous lab using scikit-learn. We'll use a LDA classifier, which we can instanciate as follows:

```python
import sklearn.discriminant_analysis as DA
Fisher=DA.LinearDiscriminantAnalysis()
```

As discussed in the lecture, to properly formulate our problem, we'll have to:

* Define the inputs (X) vs outputs (Y)
* Designate training vs testing samples (in order to get a unbias assessment of the performance of Machine Learning algorithms)

for example, here we'll take use 4M events for training and the remainder for testing.

```python
N_Train=4000000

Train_Sample=df[:N_Train]
Test_Sample=df[N_Train:]

X_Train=Train_Sample[VarNames[1:]]
y_Train=Train_Sample["signal"]

X_Test=Test_Sample[VarNames[1:]]
y_Test=Test_Sample["signal"]

Test_sig=Test_Sample[Test_Sample.signal==1]
Test_bkg=Test_Sample[Test_Sample.signal==0]
```

We can train the classifier as follow:

```python
Fisher.fit(X_Train,y_Train)
```

We can plot the output, comparing signal and background:

```python
plt.figure()
plt.hist(Fisher.decision_function(Test_sig[VarNames[1:]]),bins=100,histtype="step", color="blue", label="signal",stacked=True)
plt.hist(Fisher.decision_function(Test_bkg[VarNames[1:]]),bins=100,histtype="step", color="red", label="background",stacked=True)
plt.legend(loc='upper right')
plt.show()
```

#### Part a

Compare ROC curves computed on the test versus training samples, in a single plot. Do you see a bias?

```python
from sklearn.metrics import roc_curve, auc

# Getting scores for both training and testing
scores_train = Fisher.decision_function(X_Train)
scores_test  = Fisher.decision_function(X_Test)

# Calculating ROC curves
fpr_train, tpr_train, _ = roc_curve(y_Train, scores_train)
fpr_test,  tpr_test,  _ = roc_curve(y_Test, scores_test)

# Plotting
plt.figure(figsize=(8,6))
plt.plot(fpr_train, tpr_train, label=f"Training (AUC = {auc(fpr_train, tpr_train):.4f})")
plt.plot(fpr_test, tpr_test, label=f"Testing (AUC = {auc(fpr_test, tpr_test):.4f})")
plt.plot([0,1], [0,1], 'k--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve: Training vs Testing')
plt.legend()
plt.show()

# The training and testing ROC curves are nearly identical, with almost zero gap between their AUC values. 
# This indicates that there is no significant bias or overfitting in the LDA model. 
# Because the model is trained on a massive 4-million-event sample, it has effectively learned the underlying physics of the SUSY dataset and generalizes perfectly to the unseen testing data.
```

#### Part b

Train the Fisher performance of using the raw, features, and raw+features as input. Compare the performance one a single plot. 

```python
# Defining the sets
raw_vars = VarNames[1:9]      # First 8 physical features
derived_vars = VarNames[9:]   # The high-level features
all_vars = VarNames[1:]       # All features

sets = [("Raw", raw_vars), ("Derived", derived_vars), ("All", all_vars)]

plt.figure(figsize=(8,6))

for label, v_list in sets:
    model = DA.LinearDiscriminantAnalysis()
    model.fit(Train_Sample[v_list], y_Train)
    fpr, tpr, _ = roc_curve(y_Test, model.decision_function(Test_Sample[v_list]))
    plt.plot(fpr, tpr, label=f"{label} (AUC = {auc(fpr, tpr):.4f})")

plt.plot([0,1], [0,1], 'k--')
plt.legend()
plt.title("Performance by Feature Set")
plt.show()

# The "Raw + Derived" feature set provides the highest AUC, demonstrating that the model performs best when it has access to the full range of physical data.
# The "Derived" features (high-level variables) typically outperform the "Raw" variables alone, as they represent complex physics relationships already calculated to maximize signal separation.
# This comparison shows that while raw detector data is essential,high-level physics features are the most powerful tools for discriminating between SUSY signal and background events.
```

#### Part a
Select 3 different classifiers from the techniques listed [here](http://scikit-learn.org/stable/supervised_learning.html#supervised-learning) to compare. Note that you can use the multi-layer perceptron to build a deep network, though training may be prohibitively slow. So avoid this technique.

Write a function that takes an instantiated classifier and performs the comparison from part 3b. Use the function on your choice of functions in part a.

#### Part c

Use the best method from part c to compute the maximal significance $\sigma_S= \frac{N_S}{\sqrt{N_S+N_B}}$ for the scenarios in lab 5.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB

# We instantiate the classifiers with specific settings to manage computation time.
# n_jobs=-1 allows the Random Forest to use all available processor cores.
rf_clf = RandomForestClassifier(n_estimators=50, max_depth=10, n_jobs=-1)
dt_clf = DecisionTreeClassifier(max_depth=10)
nb_clf = GaussianNB()

print("Classifiers selected: Random Forest, Decision Tree, and Naive Bayes.")
```

```python
# Part b
def compare_performance(clf, name):
    # Training on a 500k subset to keep the lab responsive
    clf.fit(X_Train[:500000], y_Train[:500000]) 
    
    # We use predict_proba because it is standard for these types of classifiers
    # Index [:, 1] gets the probability for the 'signal' class
    probs = clf.predict_proba(X_Test)[:, 1]
    fpr, tpr, thresholds = roc_curve(y_Test, probs)
    roc_auc = auc(fpr, tpr)
    
    return fpr, tpr, roc_auc, name, thresholds

# Running the comparison for the selected models
plt.figure(figsize=(12,10))
all_results = []
models = [(rf_clf, "Random Forest"), (dt_clf, "Decision Tree"), (nb_clf, "Naive Bayes")]

for model, name in models:
    fpr, tpr, score, label, thresholds = compare_performance(model, name)
    all_results.append((fpr, tpr, score, label, thresholds))
    plt.plot(fpr, tpr, label=f"{label} (AUC = {score:.4f})")

plt.plot([0,1], [0,1], 'k--', label='Random Guess')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Performance Comparison of Different Classifiers')
plt.legend()
plt.show()
```

```python
# Part c
# We identify the best performing result from the list
best_fpr, best_tpr, _, best_name, _ = all_results[0] 

# Scenarios: (Ns, Nb)
scenarios = [(10, 100), (100, 1000), (1000, 10000), (10000, 100000)]

print(f"--- Maximal Significance Scan using {best_name} ---")

for ns, nb in scenarios:
    # Significance formula: σ = Ns / sqrt(Ns + Nb)
    # We apply the efficiencies (TPR for signal, FPR for background)
    s_values = (best_tpr * ns) / np.sqrt((best_tpr * ns) + (best_fpr * nb) + 1e-10)
    
    max_sigma = np.max(s_values)
    print(f"Scenario (Ns={ns}, Nb={nb}): Max σ = {max_sigma:.3f}")
```

Look through [model evaluation](https://scikit-learn.org/stable/modules/model_evaluation.html#) documentation. Using scikit-learns tools, compute TPR, FPR, ROC, AUC, Precision, Recall, F1 score, and accuracy for the method you selected in 4c above and each scenario. Make a nice table, which also includes the maximal significance. 


```python
# Part a
# Precision: The ability of the classifier not to label as positive a sample that is negative.
# Recall: The ability of the classifier to find all the positive samples.
# F-score: A weighted average of Precision and Recall.
# Accuracy: The fraction of predictions the model got right.

print("Metric definitions reviewed: Precision, Recall, F-score, and Accuracy.")
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score
from tabulate import tabulate

# 1. We gather the results from the best model (Random Forest)
best_model_name = all_results[0][3]
best_fpr = all_results[0][0]
best_tpr = all_results[0][1]
best_auc = all_results[0][2]

# 2. Computing hard predictions (0 or 1) using the test set
# We use the actual model object from the models list
best_clf_obj = models[0][0] 
y_pred = best_clf_obj.predict(X_Test)

# 3. Computing general performance metrics
precision = precision_score(y_Test, y_pred)
recall = recall_score(y_Test, y_pred)
f1 = f1_score(y_Test, y_pred)
accuracy = accuracy_score(y_Test, y_pred)

# 4. Preparing the table data for different scenarios
scenarios = [(10, 100), (100, 1000), (1000, 10000), (10000, 100000)]
table_rows = []

for ns, nb in scenarios:
    # Recalculating max significance for this specific scenario
    s_values = (best_tpr * ns) / np.sqrt((best_tpr * ns) + (best_fpr * nb) + 1e-10)
    max_sigma = np.max(s_values)
    
    # Each row contains the metrics and the max significance for that data size
    table_rows.append([
        f"({ns}, {nb})", 
        f"{precision:.4f}", 
        f"{recall:.4f}", 
        f"{f1:.4f}", 
        f"{accuracy:.4f}", 
        f"{best_auc:.4f}", 
        f"{max_sigma:.3f}"
    ])

# 5. Displaying the final evaluation table
headers = ["Scenario (Ns, Nb)", "Precision", "Recall", "F1 Score", "Accuracy", "AUC", "Max Sig (σ)"]
print(f"Final Model Evaluation: {best_model_name}")
print(tabulate(table_rows, headers=headers, tablefmt="grid"))
```
