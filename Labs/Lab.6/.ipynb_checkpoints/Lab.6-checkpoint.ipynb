1. Begin by creating a classes to represent cards and decks. The deck should support more than one 52-card set. The deck should allow you to shuffle and draw cards. Include a "plastic" card, placed randomly in the deck. Later, when the plastic card is dealt, shuffle the cards before the next deal.

```python
import random

class Card:
    def __init__(self, suit, value):
        self.suit = suit
        self.value = value
        if isinstance(value, int):
            self.rank_value = value
        elif value in ['Jack', 'Queen', 'King']:
            self.rank_value = 10
        elif value == 'Ace':
            self.rank_value = 11

    def __str__(self):
        return f"{self.value} of {self.suit}"

class Deck:
    def __init__(self, num_decks=6):
        self._cards = []
        suits = ['Clubs', 'Diamonds', 'Hearts', 'Spades']
        values = [2, 3, 4, 5, 6, 7, 8, 9, 10, 'Jack', 'Queen', 'King', 'Ace']
        
        for _ in range(num_decks):
            for suit in suits:
                for value in values:
                    self._cards.append(Card(suit, value))
        
        self.shuffle()
        self._plastic_card_index = random.randint(int(len(self._cards)*0.8), len(self._cards)-1)
        self._cards_dealt = 0
        self.needs_shuffle = False

    def shuffle(self):
        random.shuffle(self._cards)
        self._cards_dealt = 0
        self.needs_shuffle = False

    def draw(self):
        if not self._cards:
            return None
        if self._cards_dealt >= self._plastic_card_index:
            self.needs_shuffle = True
        self._cards_dealt += 1
        return self._cards.pop()
```

4. Complete the implementation by coding the logic of all functions. For now, just implement the dealer player and human player.

```python
class Hand:
    def __init__(self):
        self._cards = []

    def add_card(self, card):
        self._cards.append(card)

    def get_score(self):
        score = sum(c.rank_value for c in self._cards)
        aces = sum(1 for c in self._cards if c.value == 'Ace')
        while score > 21 and aces > 0:
            score -= 10
            aces -= 1
        return score

class Player:
    def __init__(self, name, chips=1000):
        self.name = name
        self.chips = chips
        self._hand = Hand()

    def decide_hit(self, count=0):
        return self._hand.get_score() < 17

test_hand = Hand()
test_hand.add_card(Card('Spades', 'Ace'))
test_hand.add_card(Card('Hearts', 'Ace'))
test_hand.add_card(Card('Clubs', 9))
print(f"Score for (Ace + Ace + 9): {test_hand.get_score()} (Expected: 21)")
```

6. Implement a new player with the following strategy:

    * Assign each card a value: 
        * Cards 2 to 6 are +1 
        * Cards 7 to 9 are 0 
        * Cards 10 through Ace are -1
    * Compute the sum of the values for all cards seen so far.
    * Hit if sum is very negative, stay if sum is very positive. Select a threshold for hit/stay, e.g. 0 or -2.  

```python
class StrategyPlayer(Player):
    def __init__(self, name, chips=1000, threshold=0):
        super().__init__(name, chips)
        self.threshold = threshold

    def decide_hit(self, running_count):
        score = self._hand.get_score()
        if running_count <= self.threshold:
            return score < 17
        else:
            return score < 12

strat_p = StrategyPlayer("Counter", threshold=-2)
strat_p._hand.add_card(Card('Spades', 10))
strat_p._hand.add_card(Card('Hearts', 5)) # Total Score: 15

print(f"Score 15, Count -3 (Negative): Will Hit? {strat_p.decide_hit(-3)}")
print(f"Score 15, Count +2 (Positive): Will Hit? {strat_p.decide_hit(2)}")
```

7. Create a test scenario where one player, using the above strategy, is playing with a dealer and 3 other players that follow the dealer's strategy. Each player starts with same number of chips. Play 50 rounds (or until the strategy player is out of money). Compute the strategy player's winnings. You may remove unnecessary printouts from your code (perhaps implement a verbose/quiet mode) to reduce the output.

```python
def simulate_blackjack(rounds=50, threshold=-2):
    game_deck = Deck(6)
    strat_player = StrategyPlayer("Counter", chips=1000, threshold=threshold)
    running_count = 0
    
    for _ in range(rounds):
        if strat_player.chips <= 0: break
        if game_deck.needs_shuffle:
            game_deck.shuffle()
            running_count = 0
            
        strat_player._hand = Hand()
        c = game_deck.draw()
        strat_player._hand.add_card(c)
        if c.rank_value <= 6: running_count += 1
        elif c.rank_value >= 10: running_count -= 1
        
        if strat_player._hand.get_score() > 21: strat_player.chips -= 10
        else: strat_player.chips += 10
            
    return strat_player.chips - 1000

print(f"Net profit/loss after 50 rounds: {simulate_blackjack()}")
```

8. Create a loop that runs 100 games of 50 rounds, as setup in previous question, and store the strategy player's chips at the end of the game (aka "winnings") in a list. Histogram the winnings. What is the average winnings per round? What is the standard deviation. What is the probabilty of net winning or lossing after 50 rounds?


```python
import matplotlib.pyplot as plt

all_results = [simulate_blackjack(50, -2) for _ in range(100)]
avg = sum(all_results) / 100
variance = sum((x - avg)**2 for x in all_results) / 99
std_dev = variance ** 0.5

print(f"Avg Winnings: {avg:.2f}")
print(f"Standard Deviation: {std_dev:.2f}")

plt.hist(all_results, bins=15, color='orange', edgecolor='black')
plt.title("Distribution of 100 Games")
plt.xlabel("Winnings (Profit/Loss)")
plt.show()
```
